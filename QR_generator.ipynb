```python
!pip install qrcode
import qrcode
import os
def generate_qr(data,filename):
  qr = qrcode.QRCode(
      version=1,
      error_correction=qrcode.constants.ERROR_CORRECT_L,
      box_size=10,
      border=4
  )
  qr.add_data(data)
  qr.make(fit=True)
  img = qr.make_image(fill_color="black",back_color="white")
  img.save(filename)
  print(f"\n QR generated succeccfully saved as {filename}")
def main():
  print("QR CODE GENERATOR:")
  # For demonstration, providing default values instead of interactive input
  data = "https://colab.research.google.com/"
  filename = "my_qr_code.png"
  generate_qr(data, filename)
if __name__ == "__main__":
  main()
```
